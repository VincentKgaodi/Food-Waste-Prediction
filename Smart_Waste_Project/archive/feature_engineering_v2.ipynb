```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/smart_waste_dataset_cleaned.csv", parse_dates=["ArrivalDate", "ExpiryDate"])
```

```python
columns_to_drop = ["DaysUntilExpiry", "SpoilageChance", "SuggestedAction"]

df = df.drop(columns=[col for col in columns_to_drop if col in df.columns])

print(f"Dropped {len(columns_to_drop)} leaky columns.")
```

#### **Date features**

```python
df["Total_Shelf_Life_Days"] = (df["ExpiryDate"] - df["ArrivalDate"]).dt.days


df = df[df["Total_Shelf_Life_Days"] >= 0]

print("Date features created.")
```

#### **Business features**

```python
df["Stock_Coverage_Ratio"] = df["StockQty"] / (df["DailySaleAvg"] + 1e-5)


df["Sales_vs_Nearby_Gap"] = df["DailySaleAvg"] - df["AvgDailySaleInNearbyStores"]

print("Business features created.")
```

#### **Boolean Conversion**

```python
df["IsSpoiled"] = df["IsSpoiled"].astype(int)
df["OnPromotion"] = df["OnPromotion"].astype(int)
df["TemperatureSensitive"] = df["TemperatureSensitive"].astype(int)

df["Zone_Missing"] = df["Zone_Missing"].astype(int)
df["TemperatureSupport_Missing"] = df["TemperatureSupport_Missing"].astype(int)

print("Booleans converted to integers.")
```

```python
missing_data = df.isna().sum()
missing_data = missing_data[missing_data > 0]

if len(missing_data) > 0:
    print("WARNING: You have missing data!\n", missing_data)
else:
    print("No missing data. Dataset is clean.")


columns_to_keep = [
    "Category", "StoreLocation", "Zone", "TemperatureSupport",
    "StockQty", "DailySaleAvg", "DistanceToNearestStore", 
    "Total_Shelf_Life_Days", "Stock_Coverage_Ratio", 
    "Sales_vs_Nearby_Gap", "AvgDailySaleInNearbyStores",
    "OnPromotion", "TemperatureSensitive", "Zone_Missing", 
    "TemperatureSupport_Missing", "IsSpoiled",
    "ArrivalDate", "ExpiryDate" 
]

final_df = df[columns_to_keep].copy()

# Save as v2 so we don't overwrite the old broken data
save_path = "../data/interim/smart_waste_features_v2.csv"
final_df.to_csv(save_path, index=False)

print(f"\nSUCCESS! Saved {len(final_df)} rows to {save_path}")
print(f"Final shape: {final_df.shape}")
```
